```python
import pandas as pd
df = pd.read_csv(r"C:\Users\lahari\OneDrive\Desktop\Dataset .csv") 
```

```python
city_counts = df['City'].value_counts()
top_city_by_restaurants = city_counts.idxmax()
top_city_count = city_counts.max()
```

```python
city_avg_rating = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False)
```

```python
top_city_by_rating = city_avg_rating.idxmax()
top_city_rating_value = city_avg_rating.max()
```

```python
print("City Analysis Results:\n")
print("City with the highest number of restaurants:\n")
print(f"{top_city_by_restaurants} — {top_city_count:,} restaurants\n")

print("Average rating for restaurants in each city (Top 5):")
print(city_avg_rating.head(5).to_frame().to_string())  # Top 5 cities

print("\nCity with the highest average rating:\n")
print(f"{top_city_by_rating} — {round(top_city_rating_value, 2)} average rating")
```
