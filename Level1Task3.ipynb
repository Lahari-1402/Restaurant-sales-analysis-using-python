```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\lahari\OneDrive\Desktop\Dataset .csv")
```

```python
price_counts = df['Price range'].value_counts().sort_index()
```

```python
total_restaurants = len(df)
price_percentages = (price_counts / total_restaurants) * 100
```

```python
plt.figure(figsize=(8, 5))
plt.bar(price_counts.index.astype(str), price_counts.values, color='skyblue')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')
plt.title('Price Range Distribution of Restaurants')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
print("Percentage of Restaurants in Each Price Range:")
print(price_percentages.round(2))
```
