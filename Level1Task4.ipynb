```python
import pandas as pd
df = pd.read_csv(r"C:\Users\lahari\OneDrive\Desktop\Dataset .csv")
```

```python
total_restaurants = len(df)
print(total_restaurants)
```

```python
online_delivery_count = (df['Has Online delivery'] == 'Yes').sum()
print(online_delivery_count)
```

```python
online_delivery_percentage = (online_delivery_count / total_restaurants) * 100
print(f"Percentage of restaurants offering online delivery: {online_delivery_percentage:.2f}%")
```

```python
avg_rating_online = df[df['Has Online delivery'] == 'Yes']['Aggregate rating'].mean()
avg_rating_offline = df[df['Has Online delivery'] == 'No']['Aggregate rating'].mean()

print(f"Average rating (Online Delivery): {avg_rating_online:.2f}")
print(f"Average rating (No Online Delivery): {avg_rating_offline:.2f}")
```
